# max(2Fe) vs freq bin — singola iniezione a SNR=8

Pulsar: `EPTA_dr2_new_wn_only` (solo rumore bianco).

Tre esperimenti, ognuno una singola iniezione per bin nei bin **7, 8, 10, 11, 14, 15**, con f0 = k / Tspan (centro bin). Per ogni bin si mostra `max(2Fe)` della mappa.

1. **ET only** vs **ET+PT+BG** — ogni iniezione calibrata a SNR=8 nella propria configurazione (ET+PT+BG usa calib-total con CRN nella covarianza).
2. **ET only** vs **ET+RN** — RN iniettato con parametri casuali per pulsar; SNR=8 sempre calibrato su ET+bianco diagonale.
3. **ET+PT+BG senza timing model in iniezione** — stessa config del punto 1 ma il GP del timing non viene iniettato; recovery invariata.

Tutte le funzioni di calibrazione, iniezione e Fe sono quelle della pipeline (`get_log10h0_for_snr_batched`, `makemodel_cgw`, `GPU_FeStat`, `build_inv_cov_per_pulsar`). Dove serve una variante del modello di iniezione (RN, no-timing) si replica la struttura di `makemodel_cgw` cambiando una sola riga.


## Setup

```python
import os
os.environ["JAX_PLATFORMS"] = "cpu"   # rimuovere se si gira su GPU

import copy
import numpy as np
import jax
import jax.numpy as jnp
import healpy as hp
import matplotlib.pyplot as plt

import discovery as ds
from injection_discovery import makemodel_cgw, makedelay_binary, compute_tref
from GPU_Fe_statistics import GPU_FeStat

# funzioni della pipeline (SNR / covarianza)
from simulator_classifier_snello_test import (
    get_log10h0_for_snr_batched,
    build_inv_cov_per_pulsar,
    get_noise_discovery,
    BG_LOG10_A, BG_GAMMA,
)

PATH = '/home/ludovicarbone/sequential_sbi_maps/feathers/EPTA_dr2_new_wn_only/'

FREQ_BINS = [7, 8, 10, 11, 14, 15]
SNR       = 8.0
NSIDE     = 16
SEED      = 345

cwcommon = ['cw_sindec', 'cw_cosinc', 'cw_log10_f0', 'cw_log10_h0',
            'cw_phi_earth', 'cw_psi', 'cw_ra', 'cw_log10_Mc']

# parametri CW fissi (angoli, massa); f0 e h0 vengono impostati per bin
BASE = dict(cw_sindec=-0.2, cw_cosinc=0.2, cw_psi=1.2, cw_phi_earth=1.5,
            cw_ra=1.35, cw_log10_Mc=9.2)

BG_PARAMS = {'crn_log10_A': BG_LOG10_A, 'crn_gamma': BG_GAMMA}
```

```python
s_dsfiles = np.sort([f for f in os.listdir(PATH) if not f.startswith('.')])
d_psrs = [ds.Pulsar.read_feather(os.path.join(PATH, f)) for f in s_dsfiles]
tref  = compute_tref(d_psrs)
tspan = ds.getspan(d_psrs)
skyloc = np.array(hp.pix2ang(NSIDE, np.arange(hp.nside2npix(NSIDE))))

print(f'{len(d_psrs)} pulsar, tspan = {tspan:.4e} s = {tspan/3.15e7:.2f} anni')
print(f'tref = {tref:.6e} s')
print('bin -> f0 [Hz]:')
for k in FREQ_BINS:
    print(f'  {k:>2d}  {k/tspan:.4e}')
```

## Funzioni comuni

`calibrate_h0` avvolge `get_log10h0_for_snr_batched` per una singola iniezione (batch di lunghezza 1). `build_inj_params` costruisce il dizionario dei parametri per `model.sample`, con la convenzione della pipeline (`d_psr`, `phi_psr`, noisedict). `fe_max` inietta, calcola la mappa e restituisce `max(2Fe)`.

```python
def calibrate_h0(psrs, f0, pulsterm, inv_covs):
    """log10_h0 per SNR=8 in un singolo bin. inv_covs=None -> bianco diagonale."""
    log10_f0 = float(np.log10(f0))
    batched = {k: np.array([BASE[k]]) for k in
               ['cw_log10_f0', 'cw_ra', 'cw_sindec', 'cw_cosinc',
                'cw_psi', 'cw_phi_earth', 'cw_log10_Mc']}
    batched['cw_log10_f0'] = np.array([log10_f0])
    log10_h0 = get_log10h0_for_snr_batched(
        psrs, batched, np.array([SNR]),
        pulsterm=pulsterm, tref=tref, inv_covs=inv_covs,
    )[0]
    return float(log10_h0)


def build_inj_params(psrs, f0, log10_h0, background):
    """Dizionario per model.sample, convenzione pipeline."""
    p = {}
    for psr in psrs:
        p.update(psr.noisedict)
        p[psr.name + '_cw_d_psr']   = 1.0
        p[psr.name + '_cw_phi_psr'] = 0.0
    p.update({k: BASE[k] for k in BASE})
    p['cw_log10_f0'] = float(np.log10(f0))
    p['cw_log10_h0'] = log10_h0
    if background:
        p.update(BG_PARAMS)
    return p


def fe_max(inj_model, rec_fstat, psrs, f0, inj_params):
    """Inietta con inj_model, recupera con rec_fstat, ritorna max(2Fe)."""
    _, residuals = inj_model.sample(jax.random.key(SEED), inj_params)
    for ii, psr in enumerate(psrs):
        psr.residuals = np.array(residuals[ii]).squeeze()
    rec_fstat._M_cache = None
    rec_fstat.precompute_M(f0)
    fe = np.array(rec_fstat.compute_Fe(f0, skyloc))
    return 2 * float(fe.max())
```

## Punto 1 — ET only vs ET+PT+BG

- **ET only**: `pulsterm=False, background=False`. SNR calibrato su bianco diagonale (`inv_covs=None`). Recovery white-noise.
- **ET+PT+BG**: `pulsterm=True, background=True`. SNR calib-total con CRN nella covarianza (`build_inv_cov_per_pulsar`). Recovery con CRN marginalizzato.

Ogni bin ha la sua iniezione indipendente, tutte a SNR=8 nella rispettiva configurazione.

```python
inv_covs_bg = build_inv_cov_per_pulsar(
    d_psrs, tspan, n_freqs=30,
    crn_log10_A=BG_PARAMS['crn_log10_A'], crn_gamma=BG_PARAMS['crn_gamma'])

max2fe_et   = []
max2fe_etpt = []

for k in FREQ_BINS:
    f0 = k / tspan

    # --- ET only ---
    psrs = copy.deepcopy(d_psrs)
    h0 = calibrate_h0(psrs, f0, pulsterm=False, inv_covs=None)
    inj = makemodel_cgw(psrs, cwcommon, pulsterm=False, background=False, tref=tref)
    rec = GPU_FeStat(psrs, lik=makemodel_cgw(psrs, cwcommon, pulsterm=False,
                                             background=False, tref=tref), tref=tref)
    ip = build_inj_params(psrs, f0, h0, background=False)
    max2fe_et.append(fe_max(inj, rec, psrs, f0, ip))

    # --- ET+PT+BG ---
    psrs = copy.deepcopy(d_psrs)
    h0 = calibrate_h0(psrs, f0, pulsterm=True, inv_covs=inv_covs_bg)
    inj = makemodel_cgw(psrs, cwcommon, pulsterm=True, background=True, tref=tref)
    rec = GPU_FeStat(psrs, lik=makemodel_cgw(psrs, cwcommon, pulsterm=True,
                                             background=True, tref=tref),
                     params=BG_PARAMS, tref=tref)
    ip = build_inj_params(psrs, f0, h0, background=True)
    max2fe_etpt.append(fe_max(inj, rec, psrs, f0, ip))

    print(f'bin {k:>2d}: ET only = {max2fe_et[-1]:8.2f}   '
          f'ET+PT+BG = {max2fe_etpt[-1]:8.2f}')
```

```python
x = np.arange(len(FREQ_BINS))
w = 0.38
fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(x - w/2, max2fe_et,   w, label='ET only',   color='#3b6fb0')
ax.bar(x + w/2, max2fe_etpt, w, label='ET+PT+BG',  color='#c0562e')
ax.set_xticks(x); ax.set_xticklabels(FREQ_BINS)
ax.set_xlabel('freq bin'); ax.set_ylabel('max(2Fe)')
ax.set_title('Punto 1 — max(2Fe) per bin, SNR=8')
ax.legend(); ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.show()
```

## Punto 2 — ET only vs ET+RN

RN iniettato con parametri casuali per pulsar (log10_A in [-14.5, -13.5], gamma in [2, 5]), scritti nel noisedict prima di costruire il modello. L'SNR resta calibrato su **ET + bianco diagonale** (`inv_covs=None`): l'RN non entra nella calibrazione, come richiesto.

Il modello di iniezione con RN e quello di recovery corrispondente non li produce `makemodel_cgw` (che non legge RN dal noisedict), quindi si replica la sua struttura aggiungendo il GP di Fourier del rumore rosso — identico al blocco usato altrove nel progetto. La recovery usa `GPU_FeStat` con `params` contenente i parametri RN, così `build_per_pulsar_noise` marginalizza il rosso.

```python
RNG = np.random.default_rng(SEED)
N_RED = 30

def draw_rn_params(psrs):
    """Scrive parametri RN casuali nel noisedict di ogni pulsar."""
    params = {}
    for p in psrs:
        log10_A = RNG.uniform(-14.5, -13.5)
        gamma   = RNG.uniform(2.0, 5.0)
        p.noisedict[p.name + '_red_noise_log10_A'] = log10_A
        p.noisedict[p.name + '_red_noise_gamma']   = gamma
        p.noisedict[p.name + '_red_components']     = N_RED
        params[p.name + '_red_noise_log10_A'] = log10_A
        params[p.name + '_red_noise_gamma']   = gamma
    return params


def make_model_rn(psrs, background=False):
    """Come makemodel_cgw ma con in piu il GP di rumore rosso per pulsar."""
    cgw_delay = makedelay_binary(pulsterm=False, tref=tref)
    pslmodels = []
    for p in psrs:
        model = [p.residuals,
                 ds.makenoise_measurement(p, p.noisedict, tnequad=True),
                 ds.makegp_timing(p, svd=True, variance=1e-40),
                 ds.makedelay(p, cgw_delay, name='cw', common=cwcommon),
                 ds.makegp_fourier(p, ds.powerlaw, N_RED, T=ds.getspan(p),
                                   name='red_noise')]
        pslmodels.append(ds.PulsarLikelihood(model))
    return ds.GlobalLikelihood(pslmodels)
```

```python
max2fe_et2  = []
max2fe_etrn = []

for k in FREQ_BINS:
    f0 = k / tspan

    # --- ET only (ricalcolato, stesso schema del punto 1) ---
    psrs = copy.deepcopy(d_psrs)
    h0 = calibrate_h0(psrs, f0, pulsterm=False, inv_covs=None)
    inj = makemodel_cgw(psrs, cwcommon, pulsterm=False, background=False, tref=tref)
    rec = GPU_FeStat(psrs, lik=makemodel_cgw(psrs, cwcommon, pulsterm=False,
                                             background=False, tref=tref), tref=tref)
    ip = build_inj_params(psrs, f0, h0, background=False)
    max2fe_et2.append(fe_max(inj, rec, psrs, f0, ip))

    # --- ET + RN ---
    psrs = copy.deepcopy(d_psrs)
    rn_params = draw_rn_params(psrs)                     # scrive nel noisedict
    h0 = calibrate_h0(psrs, f0, pulsterm=False, inv_covs=None)  # SNR su ET+bianco
    inj = make_model_rn(psrs, background=False)
    rec = GPU_FeStat(psrs, lik=make_model_rn(psrs, background=False),
                     params=rn_params, tref=tref)
    ip = build_inj_params(psrs, f0, h0, background=False)  # noisedict include RN
    max2fe_etrn.append(fe_max(inj, rec, psrs, f0, ip))

    print(f'bin {k:>2d}: ET only = {max2fe_et2[-1]:8.2f}   '
          f'ET+RN = {max2fe_etrn[-1]:8.2f}')
```

```python
x = np.arange(len(FREQ_BINS))
fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(x - w/2, max2fe_et2,  w, label='ET only', color='#3b6fb0')
ax.bar(x + w/2, max2fe_etrn, w, label='ET+RN',   color='#4a9d5b')
ax.set_xticks(x); ax.set_xticklabels(FREQ_BINS)
ax.set_xlabel('freq bin'); ax.set_ylabel('max(2Fe)')
ax.set_title('Punto 2 — max(2Fe) per bin, SNR=8 (calibrato su ET only)')
ax.legend(); ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.show()
```

## Punto 3 — ET+PT+BG senza timing model in iniezione

Stessa configurazione del punto 1 (ET+PT+BG, calib-total, SNR=8), ma il GP del timing **non viene iniettato**: si costruisce il modello di iniezione senza la riga `makegp_timing`. La recovery resta identica al punto 1 (timing marginalizzato).

Poiche in iniezione il timing GP ha `variance=1e-40`, il suo contributo ai residui e' gia' trascurabile: l'atteso e' che `max(2Fe)` cambi pochissimo rispetto al punto 1. Il confronto e' proprio questo.

```python
def make_model_notiming(psrs, pulsterm, background):
    """Come makemodel_cgw ma SENZA il GP del timing model in iniezione."""
    cgw_delay = makedelay_binary(pulsterm=pulsterm, tref=tref)
    pslmodels = []
    tspan_loc = ds.getspan(psrs)
    for p in psrs:
        model = [p.residuals,
                 ds.makenoise_measurement(p, p.noisedict, tnequad=True),
                 ds.makedelay(p, cgw_delay, name='cw', common=cwcommon)]
        pslmodels.append(ds.PulsarLikelihood(model))
    if background:
        return ds.GlobalLikelihood(
            psls=pslmodels,
            globalgp=ds.makegp_fourier_global(psrs, ds.powerlaw, ds.hd_orf,
                                              components=30, T=tspan_loc, name='crn'))
    return ds.GlobalLikelihood(pslmodels)
```

```python
max2fe_std   = []   # punto 1, ET+PT+BG (con timing in iniezione)
max2fe_notim = []   # ET+PT+BG senza timing in iniezione

for k in FREQ_BINS:
    f0 = k / tspan

    h0 = calibrate_h0(copy.deepcopy(d_psrs), f0, pulsterm=True, inv_covs=inv_covs_bg)

    # con timing (uguale al punto 1)
    psrs = copy.deepcopy(d_psrs)
    inj = makemodel_cgw(psrs, cwcommon, pulsterm=True, background=True, tref=tref)
    rec = GPU_FeStat(psrs, lik=makemodel_cgw(psrs, cwcommon, pulsterm=True,
                                             background=True, tref=tref),
                     params=BG_PARAMS, tref=tref)
    ip = build_inj_params(psrs, f0, h0, background=True)
    max2fe_std.append(fe_max(inj, rec, psrs, f0, ip))

    # senza timing in iniezione, recovery invariata
    psrs = copy.deepcopy(d_psrs)
    inj = make_model_notiming(psrs, pulsterm=True, background=True)
    rec = GPU_FeStat(psrs, lik=makemodel_cgw(psrs, cwcommon, pulsterm=True,
                                             background=True, tref=tref),
                     params=BG_PARAMS, tref=tref)
    ip = build_inj_params(psrs, f0, h0, background=True)
    max2fe_notim.append(fe_max(inj, rec, psrs, f0, ip))

    print(f'bin {k:>2d}: con timing = {max2fe_std[-1]:8.2f}   '
          f'senza timing = {max2fe_notim[-1]:8.2f}   '
          f'diff rel = {(max2fe_notim[-1]-max2fe_std[-1])/max2fe_std[-1]:+.2%}')
```

```python
x = np.arange(len(FREQ_BINS))
fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(x - w/2, max2fe_std,   w, label='ET+PT+BG (timing iniettato)',   color='#c0562e')
ax.bar(x + w/2, max2fe_notim, w, label='ET+PT+BG (no timing iniettato)', color='#8a4fbf')
ax.set_xticks(x); ax.set_xticklabels(FREQ_BINS)
ax.set_xlabel('freq bin'); ax.set_ylabel('max(2Fe)')
ax.set_title('Punto 3 — effetto del timing model in iniezione, SNR=8')
ax.legend(); ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.show()
```

## Punto 4 — ET only + iniezione a f = 1/anno esatta

Si riprendono i `max(2Fe)` ET only gia' calcolati al punto 1 (non si rifanno) e si aggiunge una singola iniezione ET only a **f0 = 1/(1 anno)** esatta, che non coincide con un centro-bin. Stessa calibrazione (SNR=8 su ET+bianco diagonale) e stessa recovery white-noise. Si mostra la mappa Fe e si aggiunge il suo `max(2Fe)` al plot del punto 1.


```python
F0_1YR = 1.0 / (365.25 * 24 * 3600)   # 1/anno in Hz
print(f'f0 (1/anno) = {F0_1YR:.4e} Hz  ->  bin equivalente = {F0_1YR*tspan:.2f}')

psrs = copy.deepcopy(d_psrs)
h0 = calibrate_h0(psrs, F0_1YR, pulsterm=False, inv_covs=None)
inj = makemodel_cgw(psrs, cwcommon, pulsterm=False, background=False, tref=tref)
rec = GPU_FeStat(psrs, lik=makemodel_cgw(psrs, cwcommon, pulsterm=False,
                                         background=False, tref=tref), tref=tref)
ip = build_inj_params(psrs, F0_1YR, h0, background=False)

# mappa completa (fe_max ricalcola solo il massimo: qui serve tutta la mappa)
_, residuals = inj.sample(jax.random.key(SEED), ip)
for ii, psr in enumerate(psrs):
    psr.residuals = np.array(residuals[ii]).squeeze()
rec._M_cache = None
rec.precompute_M(F0_1YR)
fe_1yr = np.array(rec.compute_Fe(F0_1YR, skyloc))
max2fe_1yr = 2 * float(fe_1yr.max())
print(f'max(2Fe) a 1/anno = {max2fe_1yr:.2f}')
```

```python
# mappa Fe a 1/anno, con marker della sorgente
cgw_theta = np.pi/2 - np.arcsin(BASE['cw_sindec'])
cgw_phi   = BASE['cw_ra']
fig = plt.figure(figsize=(9, 5))
hp.mollview(2*fe_1yr, rot=180, title='2Fe — ET only, f = 1/anno', hold=True)
hp.projscatter(cgw_theta, cgw_phi, marker='x', color='lime',
               s=120, linewidths=2, zorder=10)
plt.show()
```

```python
# plot del punto 1 (solo ET only) con la barra extra a 1/anno
labels = [str(k) for k in FREQ_BINS] + ['1/yr']
vals   = list(max2fe_et) + [max2fe_1yr]
colors = ['#3b6fb0']*len(FREQ_BINS) + ['#d18b2c']
x = np.arange(len(labels))
fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(x, vals, 0.6, color=colors)
ax.set_xticks(x); ax.set_xticklabels(labels)
ax.set_xlabel('freq bin'); ax.set_ylabel('max(2Fe)')
ax.set_title('Punto 4 — ET only per bin + iniezione a 1/anno, SNR=8')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.show()
```
